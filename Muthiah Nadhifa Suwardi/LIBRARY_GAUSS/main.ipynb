```python
import numpy as np
import matplotlib.pyplot as plt

from gauss_elimination.gauss_method import GaussElimination
```

```python
A1 = [
    [2, 1, -1],
    [-3, -1, 2],
    [-2, 1, 2]
]

b1 = [8, -11, -3]

gauss1 = GaussElimination(A1, b1)
gauss1.solve()
```

```python
A2 = [
    [1, 1, 1],
    [2, 3, 1],
    [1, -1, 2]
]

b2 = [6, 10, 5]

gauss2 = GaussElimination(A2, b2)
gauss2.solve()
```

Untuk Menampilkan Plot Visualisasi

```python
A3 = [
    [2, 1],
    [4, -1]
]

b3 = [5, 1]

gauss3 = GaussElimination(A3, b3)
sol = gauss3.solve()

#range x
x = np.linspace(-10, 10, 100)

#ubah ke bentuk y = mx + c
y1 = 5 - 2*x      #2x + y = 5
y2 = 4*x - 1      #4x - y =1

#plot garis
plt.plot(x, y1, label="2x + y = 5")
plt.plot(x, y2, label="4x - y = 1")

#plot titik solusi
plt.scatter(sol[0], sol[1], label="Solusi", s=100)

#untuk membatasi angka yang muncul di grafik
plt.xlim(-4, 4)
plt.ylim(-4, 4)

plt.xticks(range(-4,5))
plt.yticks(range(-4,5))

#tampilan
plt.legend()
plt.grid()
plt.title("Visualisasi SPL")
plt.xlabel("x")
plt.ylabel("y")

plt.show()
```
